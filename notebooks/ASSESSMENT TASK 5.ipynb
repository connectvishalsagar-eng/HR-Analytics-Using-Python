### 6. Create a bar chart comparing average score improvement across training programs.

```python
improvement_score_avg = df.groupby("Training Program")["improvement_score"].mean() # Calculate average score improvement for each program
print(improvement_score_avg)
```

```python
plt.figure(figsize=(12, 6), facecolor="#F3F6F9")

ax = plt.gca()
ax.set_facecolor("#FFFFFF")

programs_name = improvement_score_avg.index
programs_avg = improvement_score_avg.values

colors = ["#365F91", "#527DAA", "#6A91B5", "#78A6A6",
          "#8FAFC4", "#A3B8C8", "#648C9E", "#789B9B"]

bars = plt.bar(
    programs_name, programs_avg,
    color=colors[:len(programs_name)],
    edgecolor="white",
    linewidth=1.2,
    width=0.65
)

plt.title("Average Score Improvement Across Training Programs",
          fontsize=14, fontweight="bold", color="#263746", pad=16)

plt.xlabel("Training Programs", fontsize=11, color="#455A64")
plt.ylabel("Average Score Improvement", fontsize=11, color="#455A64")

plt.xticks(rotation=30, ha="right")
plt.yticks(color="#455A64")

plt.ylim(0, max(programs_avg) + 3)

plt.bar_label(bars, fmt="%.1f", padding=5,
              fontsize=10, fontweight="bold", color="#263746")

plt.grid(axis="y", linestyle="--", color="#D9E1E8", alpha=0.7)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```

### 8. Create a bar chart comparing training expenditure across departments

```python
training_cost = df.groupby("Department")["Training Cost"].sum()

departments = training_cost.index
cost = training_cost.values

plt.figure(figsize=(10, 7), facecolor="#F7F5F5")
ax = plt.gca()
ax.set_facecolor("#FFFFFF")

colors = ["#7D5260", "#956575", "#AD7C8B", "#C095A1",
          "#8B6878", "#B58D99", "#D0AAB2"]

bars = plt.bar(
    departments, cost,
    color=colors[:len(departments)],
    edgecolor="#FFFFFF",
    linewidth=1.2,
    width=0.65
)

plt.title("Total Training Expenditure Across Departments",
          fontsize=14, fontweight="bold", color="#49343D", pad=16)

plt.xlabel("Department", fontsize=11, color="#5C4A51")
plt.ylabel("Total Training Cost", fontsize=11, color="#5C4A51")

plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", color="#E5DDE0", alpha=0.8)
ax.set_axisbelow(True)

plt.bar_label(bars, fmt="%.0f", padding=4,
              fontsize=9, fontweight="bold", color="#49343D")

plt.tight_layout()
plt.show()
```
