# Visualization Tasks:

## Task 6: Create a Bar Chart Showing Department-wise Average Performance

In this task, we create a bar chart to compare the average performance scores of different departments. This helps us identify which department has higher and lower performance.

```python
department_avg = df.groupby("Department")["Performance Score"].mean()

plt.figure(figsize=(12, 7), facecolor="#F5F3F8")
ax = plt.gca()
ax.set_facecolor("#FFFFFF")

departments = department_avg.index
performance = department_avg.values

colors = ["#594B8A", "#7665A8", "#9384BC", "#A99BCB",
          "#7B6D9E", "#B5A9CF", "#C5BBDD"]

bars = plt.bar(
    departments, performance,
    width=0.6,
    color=colors[:len(departments)],
    edgecolor="#E5E0ED",
    linewidth=1.2
)

plt.ylim(0, max(performance) + 8)

plt.bar_label(bars, fmt="%.2f", padding=4,
              fontsize=9, fontweight="bold", color="#403653")

plt.title("Average Performance Score Across Departments",
          color="#403653", fontsize=14,
          fontweight="bold", pad=15)

plt.xlabel("Departments", fontsize=11, fontweight="bold")
plt.ylabel("Average Performance Score", fontsize=11, fontweight="bold")

plt.xticks(rotation=45, ha="right")

plt.grid(axis="y", color="#E3DDEC", linestyle="--", alpha=0.7)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```

## Task 7: Create a Scatter Plot Showing Working Hours vs Projects Completed

In this task, we create a scatter plot to see the relationship between working hours and projects completed. This helps us understand whether employees who work more hours complete more projects.

```python
hours = df["Working Hours"]  # Get working hours
projects = df["Projects Completed"]  # Get projects completed

plt.figure(figsize=(11,7), facecolor="#F3F6F9")  # Set full figure background
ax = plt.gca()
ax.set_facecolor("#FFFFFF")  # Set plot background

plt.scatter(hours, projects,
            color="#397B8A", s=60,
            marker="o", alpha=0.75,
            edgecolor="#245563", linewidth=0.7)  # Create scatter plot

plt.title("Working Hours vs Projects Completed",
          color="#263746", fontsize=14,
          fontweight="bold", pad=15)

plt.xlabel("Working Hours", fontsize=11, fontweight="bold")
plt.ylabel("Projects Completed", fontsize=11, fontweight="bold")

plt.grid(color="#D5E0E5", linestyle="--", alpha=0.7)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```

## Task 8: Create a Box Plot Showing Performance Scores across Designations

In this task, we create a box plot to compare performance scores across different designations. This helps us understand the spread of scores and identify possible outliers.

```python
ax = df.boxplot(
    column="Performance Score",
    by="Designation",
    figsize=(18,7),
    patch_artist=True,
    grid=False
)

# Apply different colors to each box
colors = ["#264653", "#2A9D8F", "#E9C46A", "#F4A261",
          "#E76F51", "#457B9D", "#8E7DBE", "#6A994E",
          "#D65DB1", "#00A6A6"]

for i, box in enumerate(ax.artists):
    box.set_facecolor(colors[i % len(colors)])
    box.set_edgecolor("#17365D")

plt.gcf().set_facecolor("#DCE6F1")
plt.gca().set_facecolor("#F4F8FC")

plt.title("Performance Scores Across Designations",
          fontsize=15, fontweight="bold", color="#17365D", pad=15)

plt.suptitle("")
plt.xlabel("Designation", fontsize=11, fontweight="bold")
plt.ylabel("Performance Score", fontsize=11, fontweight="bold")

plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.gca().set_axisbelow(True)

plt.tight_layout()
plt.show()
```

## Interpretation

From the analysis, we can see that the IT department has the highest average performance score (76.80), while Customer Support has the lowest (70.55). IT employees also have slightly more experience and complete more projects on average.

Among the designations, Managers have an average performance score of 77, while DevOps Engineers have 71.11. Employees who completed 16 projects have the highest average performance score of 79.44.

Overall, employee performance is different across departments and designations. However, completing more projects does not always mean better performance.

5. Start an Employee of the Month program to motivate employees.
6. Give bonuses and rewards to employees for their good work.
7. Encourage employees to come on time so that projects can be completed on time.
8. Encourage employees to leave work on time and spend time with their families.
